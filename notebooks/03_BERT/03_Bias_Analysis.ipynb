```python
import torch
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from transformers import AutoTokenizer, AutoModelForSequenceClassification
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
import warnings
import os
warnings.filterwarnings('ignore')

# Load model - use absolute path
model_path = r'c:\Users\aweso\Documents\fakenews-analysis\models\bert_fake_news_trained_v2'
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

tokenizer = AutoTokenizer.from_pretrained(model_path, local_files_only=True)
model = AutoModelForSequenceClassification.from_pretrained(model_path, local_files_only=True)
model.to(device)
model.eval()

print(f"✅ Model loaded from {model_path}")
print(f"✅ Running on {device}")

# Define batch prediction function
def get_predictions_batch(texts):
    """Get predictions for a batch of texts"""
    predictions = []
    probs = []
    
    for text in texts:
        inputs = tokenizer(text, truncation=True, padding=True, max_length=128, return_tensors='pt')
        inputs = {k: v.to(device) for k, v in inputs.items()}
        
        with torch.no_grad():
            outputs = model(**inputs)
            logits = outputs.logits
            prob = torch.softmax(logits, dim=1)[0].cpu().numpy()
            pred = logits.argmax(dim=1).item()
            
        predictions.append(pred)
        probs.append(prob)
    
    return np.array(predictions), np.array(probs)
```

```python

# ===== DOMAIN-SPECIFIC MODEL BIAS TEST =====
# Load political and gossip specialist models and repeat the same bias analysis

print("\n\n" + "="*80)
print("DOMAIN-SPECIFIC MODELS BIAS ANALYSIS")
print("="*80)

# Load political specialist model
print("\nLoading Political Specialist model...")
political_model_path = '../models/bert_politifact_retrained'
political_tokenizer = AutoTokenizer.from_pretrained(political_model_path)
political_model = AutoModelForSequenceClassification.from_pretrained(political_model_path)
political_model.to(device)
political_model.eval()
print("✅ Political Specialist loaded")

# Load gossip specialist model
print("Loading Gossip Specialist model...")
gossip_model_path = '../models/bert_gossip'
gossip_tokenizer = AutoTokenizer.from_pretrained(gossip_model_path)
gossip_model = AutoModelForSequenceClassification.from_pretrained(gossip_model_path)
gossip_model.to(device)
gossip_model.eval()
print("✅ Gossip Specialist loaded")

# Create batch prediction functions for domain-specific models
def get_predictions_political(texts):
    """Get predictions from political specialist model"""
    predictions = []
    probs = []
    
    for text in texts:
        inputs = political_tokenizer(text, truncation=True, padding=True, max_length=128, return_tensors='pt')
        inputs = {k: v.to(device) for k, v in inputs.items()}
        
        with torch.no_grad():
            outputs = political_model(**inputs)
            logits = outputs.logits
            prob = torch.softmax(logits, dim=1)[0].cpu().numpy()
            pred = logits.argmax(dim=1).item()
            
        predictions.append(pred)
        probs.append(prob)
    
    return np.array(predictions), np.array(probs)

def get_predictions_gossip(texts):
    """Get predictions from gossip specialist model"""
    predictions = []
    probs = []
    
    for text in texts:
        inputs = gossip_tokenizer(text, truncation=True, padding=True, max_length=128, return_tensors='pt')
        inputs = {k: v.to(device) for k, v in inputs.items()}
        
        with torch.no_grad():
            outputs = gossip_model(**inputs)
            logits = outputs.logits
            prob = torch.softmax(logits, dim=1)[0].cpu().numpy()
            pred = logits.argmax(dim=1).item()
            
        predictions.append(pred)
        probs.append(prob)
    
    return np.array(predictions), np.array(probs)

# Run bias analysis on both domain-specific models
print("\n" + "="*80)
print("SUBJECT TYPE BIAS ANALYSIS - DOMAIN-SPECIFIC MODELS (200 SYNTHETIC ARTICLES)")
print("="*80)

# Political Specialist
print("\n" + "─"*80)
print("POLITICAL SPECIALIST MODEL")
print("─"*80)

print("\nGenerating predictions on 200 articles...")
political_predictions, political_probs = get_predictions_political(df_synthetic['text'].tolist())
df_synthetic['political_pred'] = political_predictions
df_synthetic['political_confidence'] = political_probs.max(axis=1)

political_overall_acc = (political_predictions == df_synthetic['true_label']).mean()
print(f"✅ Overall accuracy: {political_overall_acc:.4f} ({political_overall_acc*100:.2f}%)")

political_results = {}
for subject_type in sorted(df_synthetic['type'].unique()):
    subset = df_synthetic[df_synthetic['type'] == subject_type]
    accuracy = (subset['political_pred'] == subset['true_label']).mean()
    
    fake_count = (subset['true_label'] == 0).sum()
    real_count = (subset['true_label'] == 1).sum()
    
    pred_fake = (subset['political_pred'] == 0).sum()
    pred_real = (subset['political_pred'] == 1).sum()
    
    fake_acc = subset[subset['true_label'] == 0]
    fake_acc = (fake_acc['political_pred'] == 0).mean() if len(fake_acc) > 0 else 0
    
    real_acc = subset[subset['true_label'] == 1]
    real_acc = (real_acc['political_pred'] == 1).mean() if len(real_acc) > 0 else 0
    
    political_results[subject_type] = {
        'accuracy': accuracy,
        'fake_acc': fake_acc,
        'real_acc': real_acc,
        'pred_fake_pct': (pred_fake / len(subset) * 100)
    }
    
    print(f"\n{subject_type.upper()}:")
    print(f"  Total articles: {len(subset)}")
    print(f"  Actual: {fake_count} fake, {real_count} real")
    print(f"  Predictions: {pred_fake} fake, {pred_real} real")
    print(f"  Overall Accuracy: {accuracy:.4f} ({accuracy*100:.2f}%)")
    print(f"  Avg confidence: {subset['political_confidence'].mean():.4f}")

# Gossip Specialist
print("\n" + "─"*80)
print("GOSSIP SPECIALIST MODEL")
print("─"*80)

print("\nGenerating predictions on 200 articles...")
gossip_predictions, gossip_probs = get_predictions_gossip(df_synthetic['text'].tolist())
df_synthetic['gossip_pred'] = gossip_predictions
df_synthetic['gossip_confidence'] = gossip_probs.max(axis=1)

gossip_overall_acc = (gossip_predictions == df_synthetic['true_label']).mean()
print(f"✅ Overall accuracy: {gossip_overall_acc:.4f} ({gossip_overall_acc*100:.2f}%)")

gossip_results = {}
for subject_type in sorted(df_synthetic['type'].unique()):
    subset = df_synthetic[df_synthetic['type'] == subject_type]
    accuracy = (subset['gossip_pred'] == subset['true_label']).mean()
    
    fake_count = (subset['true_label'] == 0).sum()
    real_count = (subset['true_label'] == 1).sum()
    
    pred_fake = (subset['gossip_pred'] == 0).sum()
    pred_real = (subset['gossip_pred'] == 1).sum()
    
    fake_acc = subset[subset['true_label'] == 0]
    fake_acc = (fake_acc['gossip_pred'] == 0).mean() if len(fake_acc) > 0 else 0
    
    real_acc = subset[subset['true_label'] == 1]
    real_acc = (real_acc['gossip_pred'] == 1).mean() if len(real_acc) > 0 else 0
    
    gossip_results[subject_type] = {
        'accuracy': accuracy,
        'fake_acc': fake_acc,
        'real_acc': real_acc,
        'pred_fake_pct': (pred_fake / len(subset) * 100)
    }
    
    print(f"\n{subject_type.upper()}:")
    print(f"  Total articles: {len(subset)}")
    print(f"  Actual: {fake_count} fake, {real_count} real")
    print(f"  Predictions: {pred_fake} fake, {pred_real} real")
    print(f"  Overall Accuracy: {accuracy:.4f} ({accuracy*100:.2f}%)")
    print(f"  Avg confidence: {subset['gossip_confidence'].mean():.4f}")

# Visualization - Compare all three models
print("\n" + "="*80)
print("COMPARING ALL THREE MODELS")
print("="*80)

fig, axes = plt.subplots(2, 2, figsize=(15, 10))

subject_types = sorted(df_synthetic['type'].unique())

# 1. Overall accuracy comparison
combined_acc = [df_synthetic[df_synthetic['type'] == st]['correct'].mean() for st in subject_types]
political_acc = [political_results[st]['accuracy'] for st in subject_types]
gossip_acc = [gossip_results[st]['accuracy'] for st in subject_types]

x = np.arange(len(subject_types))
width = 0.25

axes[0, 0].bar(x - width, combined_acc, width, label='Combined Specialist', color='#3498db')
axes[0, 0].bar(x, political_acc, width, label='Political Specialist', color='#e74c3c')
axes[0, 0].bar(x + width, gossip_acc, width, label='Gossip Specialist', color='#2ecc71')
axes[0, 0].set_ylabel('Accuracy', fontsize=11)
axes[0, 0].set_title('Overall Accuracy by Subject Type - All Models', fontsize=12, fontweight='bold')
axes[0, 0].set_xticks(x)
axes[0, 0].set_xticklabels(subject_types)
axes[0, 0].set_ylim([0, 1.05])
axes[0, 0].legend()
axes[0, 0].grid(axis='y', alpha=0.3)

# 2. Fake detection comparison
combined_fake = [results_by_type[st]['fake_acc'] for st in subject_types]
political_fake = [political_results[st]['fake_acc'] for st in subject_types]
gossip_fake = [gossip_results[st]['fake_acc'] for st in subject_types]

axes[0, 1].bar(x - width, combined_fake, width, label='Combined Specialist', color='#3498db')
axes[0, 1].bar(x, political_fake, width, label='Political Specialist', color='#e74c3c')
axes[0, 1].bar(x + width, gossip_fake, width, label='Gossip Specialist', color='#2ecc71')
axes[0, 1].set_ylabel('Detection Rate', fontsize=11)
axes[0, 1].set_title('Fake Article Detection Rate - All Models', fontsize=12, fontweight='bold')
axes[0, 1].set_xticks(x)
axes[0, 1].set_xticklabels(subject_types)
axes[0, 1].set_ylim([0, 1.05])
axes[0, 1].legend()
axes[0, 1].grid(axis='y', alpha=0.3)

# 3. Real detection comparison
combined_real = [results_by_type[st]['real_acc'] for st in subject_types]
political_real = [political_results[st]['real_acc'] for st in subject_types]
gossip_real = [gossip_results[st]['real_acc'] for st in subject_types]

axes[1, 0].bar(x - width, combined_real, width, label='Combined Specialist', color='#3498db')
axes[1, 0].bar(x, political_real, width, label='Political Specialist', color='#e74c3c')
axes[1, 0].bar(x + width, gossip_real, width, label='Gossip Specialist', color='#2ecc71')
axes[1, 0].set_ylabel('Detection Rate', fontsize=11)
axes[1, 0].set_title('Real Article Detection Rate - All Models', fontsize=12, fontweight='bold')
axes[1, 0].set_xticks(x)
axes[1, 0].set_xticklabels(subject_types)
axes[1, 0].set_ylim([0, 1.05])
axes[1, 0].legend()
axes[1, 0].grid(axis='y', alpha=0.3)

# 4. Summary statistics table
summary_data = []
for st in subject_types:
    summary_data.append({
        'Subject': st.upper(),
        'Combined': f"{results_by_type[st]['accuracy']:.1%}",
        'Political': f"{political_results[st]['accuracy']:.1%}",
        'Gossip': f"{gossip_results[st]['accuracy']:.1%}"
    })

axes[1, 1].axis('tight')
axes[1, 1].axis('off')
table = axes[1, 1].table(cellText=[[row['Subject'], row['Combined'], row['Political'], row['Gossip']] for row in summary_data],
                         colLabels=['Subject Type', 'Combined', 'Political', 'Gossip'],
                         cellLoc='center',
                         loc='center')
table.auto_set_font_size(False)
table.set_fontsize(10)
table.scale(1, 2)

plt.suptitle('Domain-Specific Models: Bias Analysis Comparison', fontsize=14, fontweight='bold', y=0.995)
plt.tight_layout()
plt.show()

print(f"\n✅ Domain-specific model analysis complete!")
```
